```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Machine learning\NLP\Emotions_Detection_NLP_ML\backend\data\train.txt",sep =';',names=["texts","emotions"],header=None)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions = df['emotions'].unique()
emotion_number = {}

i=0
for emo in unique_emotions:
    emotion_number[emo]=i
    i += 1

df['emotions'] = df['emotions'].map(emotion_number)
```

```python
df
```

```python
df['texts'] = df['texts'].str.lower()
```

```python
df.head(10)
```

```python
import string

def remove_punc(txt) :
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['texts'] = df['texts'].apply(remove_punc)
```

```python
def remove_nums(txt):
    new=''
    for i in txt:
        if not i.isdigit():
            new += i
    return new
```

```python
df['texts'] = df['texts'].apply(remove_nums)
```

```python
def remove_emojis(txt):
    new=''
    for i in txt:
        if i.isascii():
            new += i
    return new

df['texts'] = df['texts'].apply(remove_emojis)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download("punkt_tab")
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['texts']
```

```python
def remove_stop_words(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)

    return ' '.join(cleaned)
```

```python
df['texts'] = df['texts'].apply(remove_stop_words)
```

```python
df.loc[1]['texts']
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['texts'], df['emotions'], test_size=0.20, random_state=42)

```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)
```

```python
pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
print(classification_report(y_test, pred_bow))
```

```python
pred_bow
```

```python
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)


nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf,y_train)
```

```python
y_pred_tfidf = nb2_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression(max_iter=1000)
```

```python
logistic_model.fit(X_train_tfidf,y_train)
```

```python
log_pred = logistic_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, log_pred))
print(classification_report(y_test, log_pred))
```

```python
from sklearn.svm import LinearSVC, SVC

svm_model1 = LinearSVC()
svm_model1.fit(X_train_tfidf,y_train)
```

```python
svm_pred1 = svm_model1.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, svm_pred1))
print(classification_report(y_test, svm_pred1))
```

```python
svm_model2 = SVC(
    kernel="linear",
    probability=True,
    random_state=42
)
```

```python
svm_model2.fit(X_train_tfidf,y_train)
```

```python
svm_pred2 = svm_model2.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, svm_pred2))
print(classification_report(y_test, svm_pred2))
```
